# Heart Disease Prediction — Dataset Validation

## UCI Heart Disease Dataset — Cleveland Subset

### Purpose

This notebook validates the raw Cleveland subset of the UCI Heart Disease dataset before preprocessing and machine learning.

The validation includes:

- Dataset dimensions
- Column names and data types
- Missing-value detection
- Duplicate-row detection
- Original target distribution
- Unique-value validation
- Binary target construction

The raw dataset is not modified in this notebook.


```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path("../data/raw/processed.cleveland.data")

print("Dataset path:", DATA_PATH.resolve())
print("Dataset exists:", DATA_PATH.exists())

columns = [
    "age",
    "sex",
    "cp",
    "trestbps",
    "chol",
    "fbs",
    "restecg",
    "thalach",
    "exang",
    "oldpeak",
    "slope",
    "ca",
    "thal",
    "num"
]

df = pd.read_csv(
    DATA_PATH,
    names=columns,
    na_values="?"
)

print("Dataset loaded successfully.")
```

```python
print("Shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("First 5 records:")
display(df.head())

print("\nLast 5 records:")
display(df.tail())
```

```python
print("Data types:")
print(df.dtypes)
```

```python
missing_counts = df.isna().sum()

print("=== Missing-Value Report ===")
print(missing_counts[missing_counts > 0])

print("\n=== Total Missing Values ===")
print(df.isna().sum().sum())

missing_rows = df[df.isna().any(axis=1)]

print("\n=== Rows Containing Missing Values ===")
display(missing_rows)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    print("\nDuplicate rows:")
    display(df[df.duplicated(keep=False)])
else:
    print("No duplicate rows found.")
```

```python
target_counts = df["num"].value_counts().sort_index()

print("=== Original Target Distribution ===")
print(target_counts)

print("\n=== Original Target Percentages ===")
print(
    df["num"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)
```

```python
print("=== Unique Values ===")

categorical_check = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal",
    "num"
]

for column in categorical_check:
    print(f"{column}: {sorted(df[column].dropna().unique())}")
```

## Binary Target Definition

The original UCI diagnosis variable (`num`) contains five values:

- `0` — absence of heart disease
- `1–4` — presence of heart disease at different diagnostic levels

For this study, the prediction problem is formulated as binary classification.

The original `num` variable is retained for reference, while a new `target` variable is created:

- `target = 0` → No heart disease
- `target = 1` → Heart disease

```python
df["target"] = (df["num"] > 0).astype(int)

print("=== Binary Target Distribution ===")
print(df["target"].value_counts().sort_index())

print("\n=== Binary Target Percentages ===")
print(
    df["target"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)
```

```python
print("========================================")
print("FINAL DATASET VALIDATION SUMMARY")
print("========================================")

print(f"Records: {df.shape[0]}")
print("Predictor variables: 13")
print("Original target variable: num")
print("Derived target variable: target")
print(f"Total columns currently in DataFrame: {df.shape[1]}")

print(f"\nTotal missing values: {df.isna().sum().sum()}")
print(f"Number of duplicate rows: {df.duplicated().sum()}")

print("\nOriginal target:")
print(df["num"].value_counts().sort_index())

print("\nBinary target:")
print(df["target"].value_counts().sort_index())
```

### Duplicate Records

An exact duplicate-row check was performed on the 303 observations. No duplicate rows were identified.

Therefore, no duplicate records were removed from the dataset.


## Conclusion

The raw Cleveland subset of the UCI Heart Disease dataset was successfully loaded and validated.

The dataset contains 303 patient records and 13 predictor variables with the original diagnostic variable `num`. Six missing values were identified: four in `ca` and two in `thal`. The original target contains five diagnostic categories (0–4).

For the proposed binary classification problem, a derived `target` variable was created, where 0 represents absence of heart disease and 1 represents presence of heart disease.

No data imputation, duplicate removal, scaling, encoding, or model training is performed in this notebook. These operations will be addressed in subsequent preprocessing and machine-learning stages.
